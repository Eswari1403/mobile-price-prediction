## 1. Dashboard Overview

The Mobile Price Prediction Dashboard is an interactive application developed using Streamlit.

It allows users to enter mobile phone specifications such as RAM, battery power, camera specifications, and processor details.

The trained Logistic Regression model predicts the mobile phone's price category based on the entered specifications.

### Main Features
- User-friendly interface
- Mobile specification input
- Machine Learning-based prediction
- Four price categories:
  - Low Cost
  - Medium Cost
  - High Cost
  - Very High Cost

## 2. Model Information

The Logistic Regression algorithm was selected as the final model for mobile price classification.

### Model Details
- Algorithm: Logistic Regression
- Problem Type: Multi-class Classification
- Number of Classes: 4
- Validation Accuracy: 97.50%
- Model File: logistic_regression_model.pkl

The trained model was saved using Joblib and loaded into the Streamlit dashboard for making predictions.

## 3. User Input Specifications

The dashboard collects mobile phone specifications from the user.

### Important Specifications
- RAM (MB)
- Battery Power (mAh)
- Front Camera (MP)
- Primary Camera (MP)
- Number of Processor Cores
- Processor Clock Speed (GHz)

### Additional Specifications
- Bluetooth
- Dual SIM
- 3G and 4G Support
- WiFi
- Internal Memory
- Mobile Weight
- Screen Resolution
- Touch Screen
- Talk Time

## 4. Feature Engineering

Two additional features are calculated automatically from the user inputs.

### 1. Total Camera MP
Total Camera MP = Front Camera + Primary Camera

### 2. Processor Score
Processor Score = Number of Cores × Clock Speed

These engineered features are passed to the trained Logistic Regression model along with the original mobile specifications.

## 5. Prediction Process

The dashboard follows these steps to predict the mobile price category:

1. User enters mobile phone specifications.
2. Dashboard collects the entered values.
3. Total Camera MP and Processor Score are calculated.
4. The input data is converted into a DataFrame.
5. The saved Logistic Regression model processes the input.
6. The model predicts one of the four price categories.
7. The prediction result is displayed on the dashboard.

## 6. Price Categories

The Logistic Regression model classifies mobile phones into four price categories.

| Class | Predicted Category |
|---|---|
| 0 | Low Cost |
| 1 | Medium Cost |
| 2 | High Cost |
| 3 | Very High Cost |

The dashboard displays the predicted category based on the mobile specifications entered by the user.
