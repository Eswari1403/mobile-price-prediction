```python
#visualization of price distibution
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(data=train, x="price_range")
plt.title("Mobile Price Range Distribution")
plt.xlabel("Price Range")
plt.ylabel("Count")
plt.show()
```

```python
#visualization of ram distribution
plt.figure(figsize=(8, 5))

sns.histplot(data=train, x="ram", bins=30, kde=True)

plt.title("RAM Distribution")
plt.xlabel("RAM (MB)")
plt.ylabel("Frequency")
plt.show()
```

```python
#visualization of battery distribution
plt.figure(figsize=(8, 5))

sns.histplot(data=train, x="battery_power", bins=30, kde=True)

plt.title("Battery Power Distribution")
plt.xlabel("Battery Power (mAh)")
plt.ylabel("Frequency")
plt.show()
```

```python
#visualization of camera features
plt.figure(figsize=(10, 5))

sns.histplot(data=train, x="fc", color="blue", label="Front Camera", alpha=0.5)
sns.histplot(data=train, x="pc", color="orange", label="Primary Camera", alpha=0.5)

plt.title("Camera Specifications Distribution")
plt.xlabel("Camera (MP)")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```python
#Outlier Detection using Boxplots
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.boxplot(y=train["ram"])
plt.title("RAM Outliers")

plt.subplot(1, 2, 2)
sns.boxplot(y=train["battery_power"])
plt.title("Battery Power Outliers")

plt.tight_layout()
plt.show()
```

```python
#ram vs price range
plt.figure(figsize=(8, 5))

sns.boxplot(data=train, x="price_range", y="ram")

plt.title("RAM vs Price Range")
plt.xlabel("Price Range")
plt.ylabel("RAM (MB)")
plt.show()
```

```python
#battery vs price range
plt.figure(figsize=(8, 5))

sns.boxplot(data=train, x="price_range", y="battery_power")

plt.title("Battery Power vs Price Range")
plt.xlabel("Price Range")
plt.ylabel("Battery Power (mAh)")
plt.show()
```

```python
#screen resolution analysis
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=train,
    x="px_width",
    y="px_height",
    hue="price_range"
)

plt.title("Screen Resolution vs Price Range")
plt.xlabel("Pixel Width")
plt.ylabel("Pixel Height")
plt.show()
```

```python
#correlation analysis
correlation = train.corr(numeric_only=True)

print(correlation["price_range"].sort_values(ascending=False))
```

```python
#corelation heatmap
plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Feature Correlation Heatmap")
plt.show()
```

```python
#Compare Important Features with Price Range
features = ["ram", "battery_power", "fc", "pc", "n_cores", "clock_speed"]

for feature in features:
    plt.figure(figsize=(7, 4))
    sns.boxplot(data=train, x="price_range", y=feature)
    plt.title(feature + " vs Price Range")
    plt.show()
```
